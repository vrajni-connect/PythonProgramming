##product pricing dictionaries 

```python
price_dict={'mobile':100.56,'tablet':1000.67, 'featurefone':200,'laptop':2000,'desktop':567,'headset':67}
price_dict['smartwatch']=500
print(price_dict)
update_price_dict={'mobile':150.56,'tablet':1200.67, 'featurefone':250,'laptop':2200,'desktop':567,'headset':67}
price_dict.update(update_price_dict)
print(price_dict)

#remove product by name
del price_dict['headset']
print(price_dict)

#remove product if product doesntt exist
# del price_dict['headset']
if any(key=='headset' for key in price_dict):
    del price_dict['headset']
else:
    print("product not found")

price_dict.pop('headset',None)
```

## Task 4
Combined Operations

```python
product_list=['ipromax','inlight','ultrabook','chromebook','lenovo','asus']
product_list
price_dict
catalog=[(product_name, price,category) for product_name,price,category in zip(product_list,price_dict.values(),category_set)]
print(catalog)
```

```python
category_to_products ={category: [product_name for product_name, price, cat in catalog if cat == category] for category in category_set}
print(category_to_products)
print({key: len(value) for key, value in category_to_products.items()})
```
