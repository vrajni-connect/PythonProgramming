## Task 1: Discount Rules (if/else/elif)
**Objective:** Implement a discount calculation system based on order amounts using conditional statements.
- Order Amount >= $2000: 15% discount
- Order Amount >= $1500: 10% discount  
- Order Amount >= $1000: 7% discount
- Order Amount < $1000: No discount
- Tax: 5% on all orders

```python
# ========== TASK 1: Single Order Discount Calculation ==========
# Purpose: Calculate discount and final amount for a single order

# Get order amount from user
order_amount = input("Enter an order amount: ")

# Validate that the input is a valid numeric value
if order_amount.isdigit():  
    order_amount = int(order_amount)
    print(f"Order amount: ${order_amount}")
else:
    print("Invalid input. Please enter a numeric value.")
    exit()

# Calculate 5% tax on order amount
tax = 0.05 * order_amount

# Determine discount percentage based on order amount thresholds
if order_amount >= 2000:
    discount = 0.15 * order_amount  # 15% discount for large orders
elif 1500 <= order_amount < 2000:
    discount = 0.10 * order_amount  # 10% discount for medium orders
elif 1000 <= order_amount < 1500:
    discount = 0.07 * order_amount  # 7% discount for smaller orders
else:
    discount = 0  # No discount for orders below $1000

# Display results
print(f"Discount: ${discount:.2f}")
print(f"Tax: ${tax:.2f}")
print(f"Final Amount: ${order_amount - discount + tax:.2f}")
```

## Task 2: Process Multiple Orders (for loop)
**Objective:** Apply the discount rules to multiple orders using a for loop.
- Iterate through a list of orders
- Calculate discount and final amount for each order
- Display summary for each transaction

```python
# ========== TASK 2: Batch Process Multiple Orders ==========
# Purpose: Process a list of orders and calculate discounts for each

# Sample orders to process
orders = [1200, 2500, 800, 1750, 3000]

# Process each order in the list
for order_amount in orders:
    # Calculate 5% tax
    tax = 0.05 * order_amount
    
    # Determine discount percentage and amount based on order value
    if order_amount >= 2000:
        discount_percentage = 15
        discount = 0.15 * order_amount 
    elif 1500 <= order_amount < 2000:
        discount_percentage = 10
        discount = 0.10 * order_amount 
    elif 1000 <= order_amount < 1500:
        discount_percentage = 7
        discount = 0.07 * order_amount
    else:
        discount_percentage = 0
        discount = 0  # No discount for small orders
    
    # Calculate final amount after discount and tax
    final_amount = order_amount - discount + tax

    # Display order summary
    print(f'Order: ${order_amount} -> Discount%: {discount_percentage}% -> Tax: ${tax:.2f} -> Final: ${final_amount:.2f}')
    print(f'Total Revenue: ${final_amount}')
```

## Task 3: User Interactive Menu (while loop, break, continue)
**Objective:** Create an interactive menu system that allows users to:
1. Add new orders to a list
2. View all orders and their calculations
3. Quit the program
- Demonstrates: while loops, break, continue, user input validation

```python
# ========== TASK 3: Interactive Order Management System ==========
# Purpose: Menu-driven application for managing orders with while loop, break, and continue

# Initialize empty list to store orders
orders = []

# Main program loop - continues until user quits
while True:
    # Display menu options to user
    action = input("\nChoose an action (1=Add, 2=View, q=Quit): ")
    
    if action == "1":
        # ===== ADD NEW ORDER =====
        new_order = input("Enter a new order amount: ")
        
        # Validate input is a numeric value
        if new_order.isdigit():
            new_order = int(new_order)
            orders.append(new_order)  # Add to orders list
            print(f"✓ Order added: ${new_order}")
            print(f"  Current orders: {orders}")
        else:
            print("✗ Invalid input. Please enter a numeric value.")
            continue  # Skip to next iteration
    
    elif action == "2":
        # ===== VIEW ALL ORDERS WITH CALCULATIONS =====
        if orders:
            print(f"\n{'─' * 80}")
            print(f"{'Order Amount':<15} {'Discount':<15} {'Tax':<15} {'Final Amount':<15}")
            print(f"{'─' * 80}")
            
            for order_amount in orders:
                # Calculate 5% tax
                tax = 0.05 * order_amount
                
                # Calculate discount based on order amount thresholds
                if order_amount >= 2000:
                    discount = 0.15 * order_amount
                elif order_amount >= 1500:
                    discount = 0.10 * order_amount
                elif order_amount >= 1000:
                    discount = 0.07 * order_amount
                else:
                    discount = 0
                
                # Calculate final amount after discount and tax
                final_amount = order_amount - discount + tax
                
                # Display formatted order details
                print(f'${order_amount:<14.2f} ${discount:<14.2f} ${tax:<14.2f} ${final_amount:<14.2f}')
            print(f"{'─' * 80}\n")
        else:
            print("✗ No orders in the system yet.\n")
    
    elif action == "q":
        # ===== EXIT PROGRAM =====
        print("Exiting the program. Goodbye!")
        break  # Exit the while loop
    
    else:
        # ===== INVALID INPUT =====
        print("✗ Invalid action. Please choose 1, 2, or q.")
```

## Task 4: Loop Control with Conditions (break, continue)
**Objective:** Process daily sales data using break and continue statements.
- Skip zero-value days with `continue`
- Stop processing when encountering invalid (negative) values with `break`
- Accumulate valid sales amounts in a running total

```python
# ========== TASK 4: Daily Sales Summary with Loop Control ==========
# Purpose: Calculate total sales from daily data, handling invalid and zero values

# Daily sales data (some values are zero or invalid)
daily = [200, 150, 0, 400, 50, -1, 300]  
total_sales = 0

# Process each day's sales
for daily_sales_cost in daily:
    # Stop processing if negative value is encountered (data error)
    if daily_sales_cost < 0:
        print(f"✗ Invalid sales cost detected: ${daily_sales_cost} - Stopping process")
        break  # Exit the loop
    
    # Skip days with zero sales (no transactions)
    elif daily_sales_cost == 0:
        print(f"⊘ No sales on this day (${daily_sales_cost}) - Skipping")
        continue  # Go to next iteration
    
    # Process valid positive sales amounts
    elif daily_sales_cost > 0: 
        total_sales += daily_sales_cost  # Add to running total
        print(f'  Running Total: ${total_sales:<6.2f} | Added: ${daily_sales_cost}')

# Display final result
print(f"\n{'═' * 40}")
print(f"Total Sales Cost: ${total_sales}")
print(f"{'═' * 40}")
```
