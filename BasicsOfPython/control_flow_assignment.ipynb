##Discount Rules(if/else/elif)

```python
order_amount=input("enter an order amount")
if order_amount.isdigit():  
    order_amount=int(order_amount)
    print(f"order amount: {order_amount}")
else:
    print("invalid input")
    exit()

tax=0.05*order_amount
if order_amount>=2000:
    discount=0.15*order_amount 
elif 1500<=order_amount<2000:
    discount=0.10*order_amount 
elif 1000<=order_amount<1500:
    discount=0.07*order_amount
else:
    discount=0

print(f"discount: {discount}")
print(f"tax: {tax}")
print(f"final amount: {order_amount-discount}")
```
