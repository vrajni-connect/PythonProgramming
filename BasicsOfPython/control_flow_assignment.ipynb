##Discount Rules(if/else/elif)

```python
order_amount=input("enter an order amount")
if order_amount.isdigit():  
    order_amount=int(order_amount)
    print(f"order amount: {order_amount}")
else:
    print("invalid input")
    exit()

tax=0.05*order_amount
if order_amount>=2000:
    discount=0.15*order_amount 
elif 1500<=order_amount<2000:
    discount=0.10*order_amount 
elif 1000<=order_amount<1500:
    discount=0.07*order_amount
else:
    discount=0

print(f"discount: {discount}")
print(f"tax: {tax}")
print(f"final amount: {order_amount-discount+tax}")
```

## Task2 Process multiple orders foor loop

```python
orders =[1200,2500,800,1750,3000]
# 
for order_amount in orders:
    tax=0.05*order_amount
    if order_amount>=2000:
        discount_percentage=15
        discount=0.15*order_amount 
        total_revenue=order_amount-discount+tax
    elif 1500<=order_amount<2000:
        discount_percentage=10
        discount=0.10*order_amount 
        total_revenue=order_amount-discount+tax
    elif 1000<=order_amount<1500:
        discount_percentage=7
        discount=0.07*order_amount
        total_revenue=order_amount-discount+tax
    else:
        discount_percentage=0
        discount=0
        total_revenue=order_amount-discount+tax

    print(f'order amount: {order_amount}',"->",f'discount% : {discount_percentage}',"->",f'tax : {tax}',"->",f'final amount : {order_amount-discount+tax}')
    print(f'total revenue: {total_revenue}')
```

## Task4 User menu (While loop+ break continue)

```python
orders=[]
while True:
    action=input("choose an action (1=add, 2=view, q=quit): ")
    if action=="1":
        new_order=input("enter a new order amount: ")
        if new_order.isdigit():
            new_order=int(new_order)
            orders.append(new_order)
            print(f"new order added: {new_order}")
            print(f"updated orders: {orders}")
        else:
            print("invalid input")
    elif action=="2":
        if orders:
            print(f"all order amounts: {orders}")
            for order_amount in orders:
                tax=0.05*order_amount
                
                # Calculate discount based on order amount
                if order_amount>=2000:
                    discount=0.15*order_amount
                elif order_amount>=1500:
                    discount=0.10*order_amount
                elif order_amount>=1000:
                    discount=0.07*order_amount
                else:
                    discount=0
                
                final_amount=order_amount-discount+tax
                print(f'order amount: {order_amount} -> discount: {discount:.2f} -> tax: {tax:.2f} -> final amount: {final_amount:.2f}')
        else:
            print("no orders in the list")
    elif action=="q":
        print("exiting the program")
        break
    else:
        print("invalid action, please try again")
```
