## Task1

#product collection (Lists & Tuples)

```python
# Task 1: Product Collection using Lists and Tuples
# List: mutable, ordered collection of items
products=['iphone','samsung','infinix','motorola','lava','micromax']

# Tuple: immutable, ordered collection - useful for fixed data like product records
sample_product=(['iphone',1234,'mobile'],['samsung',123,'mobile'],['infinix',100,'featurephone'])

# Display types: list and tuple
print(f"Products list type: {type(products)}")
print(f"Sample product tuple type: {type(sample_product)}")
```

```python
# Demonstrate list slicing and indexing techniques
print(f"products[2]: {products[2]}")  # Access element at index 2
print(f"products[len(products)-1]: {products[len(products)-1]}")  # Last element using length
print(f"products[-1]: {products[-1]}")  # Last element using negative index
print(f"products[0:3]: {products[0:3]}")  # Slice from index 0 to 2
print(f"products[0:3:2]: {products[0:3:2]}")  # Slice with step of 2
print(f"products[::2]: {products[::2]}")  # Every 2nd element
print(f"products[::-1]: {products[::-1]}")  # Reverse the list
```

## Task2

```python
# Task 2: Working with Sets
# Create a list with duplicates, convert to set to get unique categories
category=['mobile','featurephone','tablet','laptop','desktop','mobile']
category_set=set(category)  # Set removes duplicates automatically
print(f"Unique categories (as set): {category_set}")
print(f"Number of unique categories: {len(category_set)}")
```

```python
# Check if user input category exists in the set (O(1) lookup time)
input_category=input("Enter the category: ")
if input_category in category_set:  # Set membership check is fast
    print(f"'{input_category}' found in categories: True")
else:
    print(f"'{input_category}' not found in categories: False")
```

## Task3
##product pricing dictionaries 

```python
# Task 3: Working with Dictionaries
# Create a dictionary with product categories and their prices
price_dict={'mobile':100.56,'tablet':1000.67, 'featurephone':200,'laptop':2000,'desktop':567,'headset':67}
print(f"Initial price dictionary: {price_dict}")

# Add a new product
price_dict['smartwatch']=500
print(f"\nAfter adding smartwatch: {price_dict}")

# Update multiple prices at once
update_price_dict={'mobile':150.56,'tablet':1200.67, 'featurephone':250,'laptop':2200,'desktop':567,'headset':67}
price_dict.update(update_price_dict)  # Updates existing keys, adds new ones
print(f"\nAfter price update: {price_dict}")

# Remove product by key (safe approach using pop with default)
if 'headset' in price_dict:
    price_dict.pop('headset')
    print(f"\nAfter removing headset: {price_dict}")
else:
    print("\nHeadset product not found in dictionary")
```

## Task 4
Combined Operations

```python
# Task 4: Combined Operations - Create a catalog using list comprehension
product_list=['ipromax','inlight','ultrabook','chromebook','lenovo','asus']

# Ensure data consistency: use min length to avoid index errors
# Note: category_set is unordered, so convert to list for predictable results
category_list = list(category_set)  # Convert set to list for consistent ordering
min_length = min(len(product_list), len(price_dict), len(category_list))

# Create catalog with validated data lengths
catalog = [(product_list[i], list(price_dict.values())[i], category_list[i]) 
           for i in range(min_length)]
print(f"Product Catalog:\n{catalog}")
print(f"\nNote: {len(product_list) - min_length} items were excluded due to data mismatch")
```

```python
# Group products by category using dictionary comprehension
category_to_products = {category: [product_name for product_name, price, cat in catalog if cat == category] 
                        for category in category_set}
print(f"Products grouped by category:\n{category_to_products}")

# Count products per category
product_count_by_category = {key: len(value) for key, value in category_to_products.items()}
print(f"\nProduct count by category:\n{product_count_by_category}")
```

```python
# Find and display the category with the maximum number of products
# Recalculate product count by category to ensure all variables are available
product_count_by_category = {}
for product_name, price, category in catalog:
    product_count_by_category[category] = product_count_by_category.get(category, 0) + 1

if product_count_by_category:  # Check if dictionary is not empty
    max_category = max(product_count_by_category, key=product_count_by_category.get)
    max_count = product_count_by_category[max_category]
    print(f"\n✓ Category with maximum products: '{max_category}' with {max_count} product(s)")
else:
    print("No products available in the catalog")
```
