## Task1- Basic function

```python
def apply_discount(price, discount_percentage=None):
    """
    Apply a discount to the given price.

    Parameters:
    price (float): The original price of the item.
    discount_percentage (float): The discount percentage to apply.

    Returns:
    float: The price after applying the discount.
    """
    if discount_percentage is None:
        discount_amount = price * 0.05  # Default discount is 10%
    elif discount_percentage <= 60:
        discount_amount = price * (discount_percentage / 100)
    else:
        discount_amount = 0  # No discount for 60% or more
    return price - discount_amount
```

```python
print(apply_discount(100, 10))    # Example usage
```

##Task2 Recursive function

```python
def factorial(n):
    """
    Calculate the factorial of a non-negative integer n.

    Parameters:
    n (int): A non-negative integer.

    Returns:
    int: The factorial of n.
    """
    if n < 0:
        print("Input must be a non-negative integer.")
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
factorial(5)  # Example usage
factorial(0)  # Example usage   
factorial(-3)  # Example usage
```

## Task#3 Lambda Function GST Calculator

```python
gst=lambda price: price + (price * 0.18)
```

```python
gst(100)
```

## Task#4 Using map() Apply gst to list of prices

```python
prices=[100,250,400,1200,50]
price_with_gst=list(map(gst,prices))
print(f'original prices is {prices}')
print(f'prices with GST is {price_with_gst}')
```

## Task#5 Using Filter() Filter expensive products

```python
prices=[100,250,400,1200,50,850]
list_of_prices_greater_than_500=list(filter((lambda price: price>500),prices))
list_of_prices_less_than_or_eq_500=list(filter((lambda price: price<=500),prices))
print(f'Prices greater than 500: {list_of_prices_greater_than_500}')
print(f'Prices less than or equal to 500: {list_of_prices_less_than_or_eq_500}')
```
