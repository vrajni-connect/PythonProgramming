## Task 1: Basic Function with Default Parameters
**Objective:** Create a function that applies discount to a price with optional discount percentage.
- Demonstrates: function definition, default parameters, conditional logic
- Default discount: 5% (if no percentage provided)
- Maximum allowed discount: 59% (60% or more defaults to 0% discount)

```python
# ========== TASK 1: Discount Function with Default Parameters ==========

def apply_discount(price, discount_percentage=None):
    """
    Apply a discount to the given price.
    
    Parameters:
        price (float): The original price of the item
        discount_percentage (float): Optional discount percentage to apply
                                     Defaults to 5% if not provided
    
    Returns:
        float: The final price after applying the discount
    """
    # Apply default 5% discount if no percentage is provided
    if discount_percentage is None:
        discount_amount = price * 0.05
    # Apply the specified discount if it's 59% or less
    elif discount_percentage <= 60:
        discount_amount = price * (discount_percentage / 100)
    # No discount if percentage is greater than 59%
    else:
        discount_amount = 0
    
    return price - discount_amount
```

```python
# Test the discount function with different parameters
print("Task 1: Testing apply_discount function")
print(f"Price: $100 with 10% discount = ${apply_discount(100, 10)}")  # Expected: $90
```

## Task 2: Recursive Function - Factorial Calculator
**Objective:** Calculate factorial of a number using recursion.
- Demonstrates: recursive functions, base cases, function calls within functions
- Base case: 0! = 1 and 1! = 1
- Recursive case: n! = n × (n-1)!
- Error handling: returns None for negative inputs

```python
# ========== TASK 2: Recursive Factorial Function ==========

def factorial(n):
    """
    Calculate the factorial of a non-negative integer using recursion.
    
    Parameters:
        n (int): A non-negative integer
    
    Returns:
        int: The factorial of n (n!)
        None: If input is negative (invalid)
    
    Examples:
        5! = 5 × 4 × 3 × 2 × 1 = 120
        0! = 1 (by definition)
    """
    # Check for invalid negative input
    if n < 0:
        print("✗ Input must be a non-negative integer.")
        return None
    
    # Base case: factorial of 0 or 1 is 1
    elif n == 0 or n == 1:
        return 1
    
    # Recursive case: n! = n × (n-1)!
    else:
        return n * factorial(n - 1)
```

```python
# Test the factorial function
print("\nTask 2: Testing factorial function")
print(f"factorial(5) = {factorial(5)}")      # Expected: 120
print(f"factorial(0) = {factorial(0)}")      # Expected: 1
print(f"factorial(-3) = {factorial(-3)}")    # Expected: None with error message
```

## Task 3: Lambda Function - GST Calculator
**Objective:** Create an anonymous function using lambda to calculate price with GST (Goods and Service Tax).
- Demonstrates: lambda functions, anonymous functions
- Calculation: Price + (Price × 18%) where 18% is the GST rate
- Format: `lambda parameter: expression`

```python
# ========== TASK 3: Lambda Function for GST Calculation ==========
# GST (Goods and Service Tax) rate = 18%

gst = lambda price: price + (price * 0.18)

# Lambda syntax explanation:
# - lambda: keyword for anonymous function
# - price: input parameter
# - price + (price * 0.18): expression that calculates final price with GST
```

```python
# Test the GST lambda function
result = gst(100)
print(f"\nTask 3: Testing GST lambda function")
print(f"Price: $100 with 18% GST = ${result}")  # Expected: $118
```

## Task 4: Using map() - Apply GST to List of Prices
**Objective:** Use the `map()` function to apply a function to each item in a list.
- Demonstrates: map() function, functional programming, lambda with map
- `map()` iterates through each element and applies the function
- Returns a map object (converted to list here)

```python
# ========== TASK 4: Using map() to Apply GST to Multiple Prices ==========

prices = [100, 250, 400, 1200, 50]

# Use map() to apply the gst lambda function to each price in the list
price_with_gst = list(map(gst, prices))

# Display original and GST-applied prices
print(f"\nTask 4: Using map() function")
print(f"Original prices:      {prices}")
print(f"Prices with GST (18%): {price_with_gst}")
```

## Task 5: Using filter() - Filter Products by Price
**Objective:** Use the `filter()` function to separate prices into two categories.
- Demonstrates: filter() function, filtering with lambda, conditional logic
- Splits prices into: expensive (>$500) and affordable (≤$500)
- Returns a filter object (converted to list here)

```python
# ========== TASK 5: Using filter() to Categorize Prices ==========

prices = [100, 250, 400, 1200, 50, 850]

# Filter prices greater than $500 (expensive items)
list_of_prices_greater_than_500 = list(filter((lambda price: price > 500), prices))

# Filter prices less than or equal to $500 (affordable items)
list_of_prices_less_than_or_eq_500 = list(filter((lambda price: price <= 500), prices))

# Display the filtered results
print(f"\nTask 5: Using filter() function")
print(f"Original prices:                {prices}")
print(f"Prices > $500 (Expensive):      {list_of_prices_greater_than_500}")
print(f"Prices ≤ $500 (Affordable):     {list_of_prices_less_than_or_eq_500}")
```

## Task 6: Combined Utility Function - map() and filter() Together
**Objective:** Combine map() and filter() to process a list of prices.
- Demonstrates: combining functional programming tools
- Step 1: Apply 10% discount to all prices using map()
- Step 2: Filter prices greater than $300 using filter()

```python
# ========== TASK 6: Combined Utility Function (map + filter) ==========

def process_prices(prices):
    """
    Process a list of prices by applying discount and filtering high-value items.
    
    Parameters:
        prices (list): List of price values
    
    Returns:
        tuple: (discounted_prices, filtered_high_value_prices)
               - discounted_prices: All prices with 10% discount applied
               - filtered_high_value_prices: Prices > $300 after discount
    """
    # Step 1: Apply 10% discount to all prices using map()
    discount_10_percent = list(map(lambda price: price - (price * 0.10), prices))
    
    # Step 2: Filter to keep only prices greater than $300 using filter()
    discounted_prices_greater_than_300 = list(filter(lambda price: price > 300, discount_10_percent))
    
    return discount_10_percent, discounted_prices_greater_than_300

# Test the combined function
print(f"\nTask 6: Combined map() and filter()")
test_prices = [100, 500, 900, 50, 750]
discounted, filtered = process_prices(test_prices)
print(f"Original prices:                    {test_prices}")
print(f"After 10% discount:                 {discounted}")
print(f"Items > $300 after discount:        {filtered}")
```

## Task 7: Interactive Menu System Using Functions
**Objective:** Create an interactive menu that uses helper functions for price management.
- Demonstrates: function composition, menu-driven programming
- Functions:
  1. `add_price()` - Add a new price to the list
  2. `get_average_price()` - Calculate average of all prices
  3. `get_max_price()` - Find the maximum price

```python
# ========== TASK 7: Interactive Menu System Using Functions ==========

def add_price(price_list, price):
    """
    Add a new price to the price list.
    
    Parameters:
        price_list (list): Current list of prices
        price (float): New price to add
    
    Returns:
        list: Updated list with the new price appended
    """
    return price_list + [price]

def get_average_price(price_list):
    """
    Calculate the average price from the list.
    
    Parameters:
        price_list (list): List of prices
    
    Returns:
        float: Average price (0 if list is empty)
    """
    if len(price_list) == 0:
        return 0
    return sum(price_list) / len(price_list)

def get_max_price(price_list):
    """
    Find the maximum price in the list.
    
    Parameters:
        price_list (list): List of prices
    
    Returns:
        float: Maximum price (None if list is empty)
    """
    if len(price_list) == 0:
        return None
    return max(price_list)

# Interactive menu
print(f"\nTask 7: Interactive Menu System")
print("=" * 50)

menu_options = input(
    "Choose an option:\n"
    "  1. Add a price\n"
    "  2. Get average price\n"
    "  3. Get maximum price\n"
    "  q. Quit\n"
    "Enter your choice: "
)

# Handle user selection
if menu_options == '1':
    price = float(input("Enter the price to add: "))
    price_list = []
    price_list = add_price(price_list, price)
    print(f"✓ Price ${price} added. Current list: {price_list}")

elif menu_options == '2':
    price_list = [100, 250, 400, 1200, 50]
    average_price = get_average_price(price_list)
    print(f"Price list: {price_list}")
    print(f"✓ Average price: ${average_price:.2f}")

elif menu_options == '3':
    price_list = [100, 250, 400, 1200, 50]
    max_price = get_max_price(price_list)
    print(f"Price list: {price_list}")
    print(f"✓ Maximum price: ${max_price}")

elif menu_options.lower() == 'q':
    print("Exiting the program.")

else:
    print("✗ Invalid option. Please choose 1, 2, 3, or q.")
```
