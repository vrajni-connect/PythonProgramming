## Task#3- Append New Sales 

```python
# Task 3: Append new sale values to the existing file and display all values.
new_sales = [5000, 2500, 1700]
sales_data_file = open('sales_data.txt', 'a')
for new_sale in new_sales:
    sales_data_file.write(str(new_sale) + '\n')
sales_data_file.close()
# Verify the contents by reading the file.
read_sales_data_file = open('sales_data.txt', 'r')
for line in read_sales_data_file:
    print(line.strip())

read_sales_data_file.close()
```

## Task#4 - Generate Summary report from file

```python
# Task 4: Generate a summary report (total, average, highest, lowest) from the sales file.
read_all_sales_values = open('sales_data.txt', 'r')

total_sale = 0
highest_sale = None
lowest_sale = None
count = 0

for sale in read_all_sales_values:
    # Convert each line to integer and handle numeric computation.
    value = int(sale.strip())
    total_sale += value
    count += 1
    if highest_sale is None or value > highest_sale:
        highest_sale = value
    if lowest_sale is None or value < lowest_sale:
        lowest_sale = value

read_all_sales_values.close()

print("total sale:", total_sale)
print("average sale:", total_sale / count if count else 0)

print("highest sale:", highest_sale)print("lowest sale:", lowest_sale)
```
