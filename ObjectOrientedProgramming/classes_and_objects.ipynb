## Task 1 - create class product with attributes name, price & category

```python
class Product:
    def __init__(self,name, price, category):
        self.name=name
        self.__price=price
        self.category=category

    def get_info(self):
        print(f"product name: {self.name}, price: {self.get_price()}, category: {self.category}")

    def get_price(self):
        return self.__price
    
    def set_price(self,new_price):
        if new_price>0:
            self.__price=new_price
            print(f"New price for {self.name} is {self.__price}")
    
    def __str__(self):
        return f"Product(name={self.name}, price={self.get_price()}, category={self.category})" 

    def __add__(self, other):
        if isinstance(other, Product):
            return self.get_price() + other.get_price()
        
        
```

```python
product=Product("laptop", 1000, "electronics")
product.get_info()
product2=Product("phone", 500, "electronics")
product2.get_info()
```

```python
product1=Product("laptop", 1000, "electronics")
print(product1.get_price())
product1.set_price(1200)
print(product1.get_price())
```

## Task3 - Inheritance(Single Level)

```python
class ElectronicProduct(Product):
    def __init__(self, name, price, category, warranty_years):
        super().__init__(name, price, category)
        self.warranty_years=warranty_years
    def get_info(self):
        super().get_info()
        print(f"warranty years: {self.warranty_years}")
```

```python
electronic_product=ElectronicProduct("laptop", 1000, "electronics", 2)
electronic_product.get_info()
```

## Task4-Polimorphism

```python
class Laptop(Product):
    def __init__(self, name, price, category, brand):
        super().__init__(name, price, category)
        self.brand=brand
    
    def get_info(self):
        super().get_info()
        print(f"brand: {self.brand}")

laptop=Laptop("laptop", 1000, "electronics", "Dell")
laptop.get_info()
```

```python
class mobile(Product):
    def __init__(self, name, price, category, brand):
        super().__init__(name, price, category)
        self.brand=brand
    
    def get_info(self):
        super().get_info()
        print(f"brand: {self.brand}")

mobile=mobile("phone", 500, "electronics", "Samsung")
mobile.get_info()
```

```python
## write a loop that iterates over objects of Laptop & Mobile and calls get_info() method for each object
## to show polimorphism in action

products=[laptop, mobile]
for product in products:
    product.get_info()

    
```

## Task - 5 Abstraction - using abstract base class


```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print(f"Processing credit card payment of {amount}")    


class UPIpayment(Payment):
    def process_payment(self, amount):
        print(f"Processing UPI payment of {amount}")

cc = CreditCardPayment()
cc.process_payment(100)
upi = UPIpayment()
upi.process_payment(100)    

payment_methods=[CreditCardPayment(), UPIpayment()]
for payment in payment_methods:
    payment.process_payment(100)

```

## Task7 - Mini Project Simple Inventory system (OOP only)

```python
class Inventory:

    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, product):
        self.products.remove(product)

    def show_all_products(self):
        for product in self.products:
            print(f"- {product.name}: ${product.get_price()}")

    def __add__(self, other):
        if isinstance(other, Inventory):
            new_inventory = Inventory()
            new_inventory.products = self.products + other.products
            return new_inventory
        
    def get_total_value(self):
        total = 0
        for product in self.products:
            total += product.get_price()
        return total


class Store:
    def __init__(self, store_name):
        self.name=store_name
        self.inventory=Inventory()
    
    def add_new_product(self, product):
        name=input("Enter product name: ")
        price=float(input("Enter product price: "))
        category=input("Enter product category: ")
        product = Product(name, price, category)
        self.inventory.add_product(product)

    def show_summary(self):
        print(f"Store name: {self.name}")
        self.inventory.show_all_products()
        print(f"Total inventory value: {self.inventory.get_total_value()}")
```
